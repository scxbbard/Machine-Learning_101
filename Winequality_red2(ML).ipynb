```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_predict, KFold
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.linear_model import LinearRegression, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import statsmodels.api as sm
```

```python
# Load dataset
df = pd.read_csv('winequality-red.csv', sep=';')
X = df.drop('quality', axis=1).values
y = df['quality'].values
```

```python
# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
```

```python
# Cross-validation setup
kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
# KNN
knn = KNeighborsRegressor(n_neighbors=5)
knn.fit(X_train, y_train)
y_pred_knn = knn.predict(X_test)
print("KNN Holdout:", mean_squared_error(y_test, y_pred_knn),
      mean_absolute_error(y_test, y_pred_knn),
      r2_score(y_test, y_pred_knn))

y_pred_knn_cv = cross_val_predict(knn, X_scaled, y, cv=kf)
print("KNN CV:", mean_squared_error(y, y_pred_knn_cv),
      mean_absolute_error(y, y_pred_knn_cv),
      r2_score(y, y_pred_knn_cv))
```

```python
# Linear Regression
lin = LinearRegression()
lin.fit(X_train, y_train)
y_pred_lin = lin.predict(X_test)
print("Linear Holdout:", mean_squared_error(y_test, y_pred_lin),
      mean_absolute_error(y_test, y_pred_lin),
      r2_score(y_test, y_pred_lin))

y_pred_lin_cv = cross_val_predict(lin, X_scaled, y, cv=kf)
print("Linear CV:", mean_squared_error(y, y_pred_lin_cv),
      mean_absolute_error(y, y_pred_lin_cv),
      r2_score(y, y_pred_lin_cv))

print("-----------------------------------------")
lin.summary()
```

```python
# Decision Tree
tree = DecisionTreeRegressor(random_state=42)
tree.fit(X_train, y_train)
y_pred_tree = tree.predict(X_test)
print("DecisionTree Holdout:", mean_squared_error(y_test, y_pred_tree),
      mean_absolute_error(y_test, y_pred_tree),
      r2_score(y_test, y_pred_tree))

y_pred_tree_cv = cross_val_predict(tree, X_scaled, y, cv=kf)
print("DecisionTree CV:", mean_squared_error(y, y_pred_tree_cv),
      mean_absolute_error(y, y_pred_tree_cv),
      r2_score(y, y_pred_tree_cv))
```

```python
# Random Forest
forest = RandomForestRegressor(n_estimators=100, random_state=42)
forest.fit(X_train, y_train)
y_pred_forest = forest.predict(X_test)
print("RandomForest Holdout:", mean_squared_error(y_test, y_pred_forest),
      mean_absolute_error(y_test, y_pred_forest),
      r2_score(y_test, y_pred_forest))

y_pred_forest_cv = cross_val_predict(forest, X_scaled, y, cv=kf)
print("RandomForest CV:", mean_squared_error(y, y_pred_forest_cv),
      mean_absolute_error(y, y_pred_forest_cv),
      r2_score(y, y_pred_forest_cv))
```

```python
# Lasso Regression
lasso = Lasso(alpha=0.1, random_state=42)
lasso.fit(X_train, y_train)
y_pred_lasso = lasso.predict(X_test)
print("Lasso Holdout:", mean_squared_error(y_test, y_pred_lasso),
      mean_absolute_error(y_test, y_pred_lasso),
      r2_score(y_test, y_pred_lasso))

y_pred_lasso_cv = cross_val_predict(lasso, X_scaled, y, cv=kf)
print("Lasso CV:", mean_squared_error(y, y_pred_lasso_cv),
      mean_absolute_error(y, y_pred_lasso_cv),
      r2_score(y, y_pred_lasso_cv))
```
